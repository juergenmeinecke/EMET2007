## Summary

### Key Takeaways

1. **Always visualise your data** before running regressions
2. **Outliers can have large influence** on OLS estimates (because we minimise squared errors)
3. **Check sensitivity** of your results by re-running without outliers
4. **Report both results** when outliers substantially change conclusions
5. **Think carefully** before removing outliers - is there a substantive reason, or are you just data-mining?

### What Happened Here?

- With Malta: The relationship was statistically significant (p < 0.05)
- Without Malta: The relationship was no longer significant at the 5% level
- Malta, being an extreme outlier on the x-axis, had **leverage** that pulled the regression line toward it
- This is a cautionary tale about relying on a single outlier for inference
