## Part 4: Multiple Mutually Exclusive Dummies

### The Dummy Variable Trap

We have four dummy variables for prenatal visits:

| Variable | Description |
|:---------|:------------|
| `tripre0` | No prenatal visits |
| `tripre1` | First visit in 1st trimester |
| `tripre2` | First visit in 2nd trimester |
| `tripre3` | First visit in 3rd trimester |

**Important:** These are **mutually exclusive** and **exhaustive** - every observation belongs to exactly one category.

**Rule:** Always omit one category as the **reference category**. The coefficients on included dummies show differences relative to this reference.

## Summary

### Key Concepts

**Dummy Variables in Regression:**
- Intercept = mean of Y when dummy = 0 (reference group)
- Coefficient = difference in means between dummy = 1 and dummy = 0
- Same test statistic as a two-sample t-test

**Multiple Mutually Exclusive Dummies:**
- Always omit one category (reference category)
- Coefficients show differences relative to the reference
- Including all dummies causes perfect multicollinearity ("dummy variable trap")

**Causal Interpretation:**
- Observational data makes causal interpretation difficult
- Smoking is a choice variable - may be correlated with other factors
- Controls help but cannot fully address selection bias
- Be cautious about interpreting coefficients as causal effects
