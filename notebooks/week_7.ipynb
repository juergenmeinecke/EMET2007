## Important Note

**From now on, always use heteroskedasticity-robust standard errors!**

In practice, we rarely know if homoskedasticity holds. Robust standard errors protect against this uncertainty at little cost when homoskedasticity does hold.

## Part 4: Joint Hypothesis Testing (F-test)

### Exercise 7: Test Joint Significance of Control Variables

Sometimes we want to test whether **multiple** coefficients are jointly equal to zero. This requires an F-test.

**Null hypothesis:**
$$H_0: \beta_2 = \beta_3 = \beta_4 = \beta_5 = 0$$

This tests whether we can exclude ALL the control variables from the regression.

**Syntax:**

## Summary

### Key Concepts

**Heteroskedasticity-Robust Standard Errors:**
- Use `cov_type='HC1'` in statsmodels
- Always use robust SEs in practice
- Coefficients stay the same; only SEs/t-stats/p-values change

**Multiple Regression:**
- Controls for other factors that might affect Y
- Coefficients represent partial effects (holding other variables constant)
- Helps address omitted variable bias

**Joint F-tests:**
- Test whether multiple coefficients are jointly zero
- Useful for testing groups of variables
- Reject if p-value < significance level

### Key Findings

- The effect of trade share on growth became smaller and less significant after adding controls
- This suggests the simple regression suffered from omitted variable bias
- The control variables are jointly significant (we can't omit them all)
