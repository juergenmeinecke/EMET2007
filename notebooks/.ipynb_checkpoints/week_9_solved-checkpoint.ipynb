Econometrically, this exercise uses interaction terms between a continuous and a binary variable to elicit the effect of lead on mortality. 

## Imports and loading data

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import statsmodels.stats.api as sms
import statsmodels.formula.api as smf
```

```python
# from google.colab import drive
# drive.mount('/content/drive')
# df = pd.read_csv('drive/MyDrive/EMET2007/datasets/lead_mortality.csv')
```

```python
df = pd.read_csv('../datasets/lead_mortality.csv')
```

```python
def t_test(x1, x2):
    numerator = x1.mean() - x2.mean()  # aka point estimate
    se1 = x1.std() / np.sqrt(len(x1))
    se2 = x2.std() / np.sqrt(len(x2))
    denominator = np.sqrt(se1**2 + se2**2)
    t_stat = numerator / denominator  # our t statistic
    ci_lb = numerator - 1.96 * denominator  # lower bound
    ci_ub = numerator + 1.96 * denominator  # upper bound
    ci = (ci_lb, ci_ub)
    print('Two-sample t-test')
    print(f'Mean in group 1: {x1.mean()}')
    print(f'Mean in group 2: {x2.mean()}')
    print(f'Point estimate for difference in means: {numerator}')
    print(f'Test statistic: {t_stat}')
    print(f'95% confidence interval: {ci}')
    return numerator, t_stat, ci
```

## Exercise 3


The amount of lead leached from lead pipes depends on the chemistry of the water running through the
pipes. The more acidic the water is (lower pH) the more lead is leached. It seems natural to study
if cities with lead pipes **and** low pH levels have particularly high infant mortality rate. The
way to do this in a regression analysis is by using interaction terms.

Plot a histogram of ``ph`` to get a sense of the distribution of water acidity.


```python
fig, ax = plt.subplots()

ax.hist(df.ph, ec='black', bins=np.arange(5, 10, 0.5))
ax.set_xlabel('pH')
ax.set_ylabel('Frequency')
ax.set_title('Histogram of pH levels')
plt.show()
```

## Exercise 4 

### Exercise 4a
Run the following alternative regressions:

* specification 1: ``infRate`` on ``lead``
* specification 2: ``infRate`` on ``lead`` and ``ph``
* specification 3: ``infRate`` on ``lead``, ``ph``, and the interaction term ``lead`` x ``ph``.


```python
formula1 = 'infRate ~ lead'
reg1 = smf.ols(formula1, data=df, 
                missing='drop').fit(cov_type='HC1', use_t=False)
reg1.summary()
```

```python
formula2 = 'infRate ~ lead + ph'
reg2 = smf.ols(formula2, data=df, 
                missing='drop').fit(cov_type='HC1', use_t=False)
reg2.summary()
```

```python
formula3 = 'infRate ~ lead*ph'
reg3 = smf.ols(formula3, data=df, 
               missing='drop').fit(cov_type='HC1', use_t=False)
reg3.summary()
```


### Exercise 4b
Discuss the differences between each specification. Carefully discuss all coefficients in specification 3!

Specification 1 is a simple linear regression that attempts to estimate the effect of lead on infrate.  The slope coefficient estimate for lead is not significant at the 5% significance level.

Specification 2 adds the ph variable to the regression in Specification 1. Interestingly, the coefficient estimates for both lead and ph are now significant, implying that they both have an effect on infant mortality.

However, we do not suspect that pH alone has any effect on infant mortality, so to see the whole story here we need to add the interaction term.

Specification 3 adds an interaction term between lead and ph. This allows us to account for differing pH levels in studying whether lead pipes have an effect on infant mortality.

Specification 3 has two associated population regression functions: one when lead = 0 and one when lead = 1. The four coefficients under Specification 3 are:
- $\beta_0$, the intercept
- $\beta_1$, the coefficient on lead
- $\beta_2$, the coefficient on ph
- $\beta_3$, the coefficient for the interaction term

When lead = 0:
- the intercept of the estimated PRF = $\hat{\beta_0}$
- the slope of the estimated PRF = $\hat{\beta_2}$

When lead = 1:
- the intercept of the estimated PRF = $\hat{\beta_0} + \hat{\beta_1}$
- the slope of the estimated PRF = $\hat{\beta_2} + \hat{\beta_3}$

Therefore, for the two estimated PRFs:
- difference in slope: $\hat{\beta_3}$
- difference in intercept: $\hat{\beta_1}$


## Exercise 5 

Using specification 3, does ``lead`` have a statistically significant effect on ``infRate``?


```python
ftest = reg3.f_test('lead = lead:ph = 0')
ftest
```

p-value = 0.0214
=> `lead` has a statistically significant effect on `infRate` at the 5% significance level.

## Exercise 7
    
### Exercise 7a
Write down the
* estimated PRF for a city with ``lead=1``
* estimated PRF for a city with ``lead=0``
* difference between both estimated PRFs


```python
# Estimated PRF for city with non-lead pipes:
prfhat0 = '$$ \\widehat{PRF_0} =' 
prfhat0 += f' {round(reg3.params[0], 4)} + '
prfhat0 += f'{round(reg3.params[2], 4)}ph $$'
# Estimated PRF for city with lead pipes:
prfhat1 = '$$ \\widehat{PRF_1} =' 
prfhat1 += f' {round(reg3.params[0] + reg3.params[1], 4)} + '
prfhat1 += f'{round(reg3.params[2] + reg3.params[3], 4)}ph $$'
# Difference in estimated PRFs:
diff_prfs = '$$ \\widehat{PRF_1} - \\widehat{PRF_0} =' 
diff_prfs += f' {round(reg3.params[1], 4)} + {round(reg3.params[3], 4)}ph $$'
print(reg3.params)
print(prfhat0)
print(prfhat1)
print(diff_prfs)
```

$$ \widehat{PRF_0} = 0.9189 + -0.0752ph $$
$$ \widehat{PRF_1} = 1.3807 + -0.132ph $$
$$ \widehat{PRF_1} - \widehat{PRF_0} = 0.4618 + -0.0569ph $$


### Exercise 7b
All three are functions in ``ph``. Produce a scatter plot of ``infRate`` versus ``ph`` and add the three functions to the graph.  Use this graph to study how the effect of lead pipes changes as the pH level changes. 
    

```python
x = np.linspace(5, 9, 10)
fig, ax = plt.subplots()

prfhat0 = reg3.params[0] + reg3.params[2]*x
prfhat1 = reg3.params[0] + reg3.params[1] + (reg3.params[2] + reg3.params[3])*x
diff_prfs = reg3.params[1] + reg3.params[3]*x
ax.plot(df.ph, df.infRate, '.', color = 'black')
ax.plot(x, prfhat0, color = 'green', 
        label='Estimated PRF (no lead)')
ax.plot(x, prfhat1, color = 'red', 
        label='Estimated PRF (lead)')
ax.plot(x, diff_prfs, color = 'blue', 
        label='Difference in estimated PRFs')
ax.set_xlabel('pH')
ax.set_ylabel('Infant mortality rate (deaths per 100)')
ax.set_title('Infant mortality rate vs pH')
ax.legend(loc='upper right', 
          bbox_to_anchor=(1.3, 1))
ax.grid()
plt.show()
```

**Observations:**

Upon inspecting this plot, it appears that lead pipes are likely to have a more severe effect on infant mortality when the pH level of the city's water is low.  This can be seen by noting that the estimated difference between the estimated PRFs (the light blue line) decreases as the pH level increases.

Cities with no lead: smoother slope and initial value (intercept) is lower

Cities with lead pipes: shift up of the intercept, and the slope becomes steeper

We can observe that the difference between PRFs gets lower as pH increases. Where the PRFs intersect, the difference equals zero and then becomes negative as pH continues increasing.
