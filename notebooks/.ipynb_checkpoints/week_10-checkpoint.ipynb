```python
df = pd.read_csv('../datasets/cpi_aus_2025.csv')
```

Run this cell once you've loaded the data:

```python
df['date'] = pd.to_datetime(df['time'], format='%b-%Y')
df.index = pd.DatetimeIndex(df.date, name='quarter').to_period('Q')
```

## Exercise 2

Plot both time series. Look what happened to inflation over time!


```python
fig, axs = plt.subplots(2, 1, figsize=(8,7))

axs[0].plot(df.date, df.cpi)
axs[0].set_ylabel('CPI')
axs[0].set_title('CPI (Australia)')
axs[1].plot(df.date, df.infl)
axs[1].set_ylabel('Inflation (annualised, %)')
axs[1].set_title('Inflation (Australia)')
fig.supxlabel('Time')
plt.show()
```
