## Exercise 3

Add a new (categorical) variable ``wabove`` to your data frame. The definition of ``wabove`` is as follows: it equals 1 if ``weightkg`` exceeds 73.9 kilograms (it is **above** the median weight in the original sample), and zero otherwise.
</div>

```python
df_clean['wabove'] = df_clean.weightkg.apply(lambda x: 1 if x > 73.9 else 0)
```

<div class="ex">
    
## Exercise 4


### Exercise 4a
For the two sub-samples that are created by ``wabove``, collect the following information for ``earnings``:


|  Information | ``wabove == 0`` | ``wabove == 1`` |
|:--|:--|:--|
| Sample size *n* | | |
| Average | | | 
| Standard deviation | | |
| Standard error | | |
    
</div>

```python
df_n = df_clean.groupby('wabove').earnings.count().rename('n')
df_mean = df_clean.groupby('wabove').earnings.mean().rename('average')
df_sd = df_clean.groupby('wabove').earnings.std().rename('sd')
df_se = df_clean.groupby('wabove').earnings.sem().rename('se')
```

```python
results_df = pd.concat((df_n, df_mean, df_sd, df_se), axis=1)
results_df
```

<div class="ex">
    
Construct a 95% confidence interval for the earnings difference between the two groups. Do it in two different ways:

### Exercise 4b
*   The hard way: Implement the appropriate statistical test by programming in ``Python``. [1 mark]

</div>

```python
diff_avg_earnings = results_df.average[1] - results_df.average[0]
se_diff_avg_earnings = np.sqrt(results_df.se[0]**2 + results_df.se[1]**2)
t_stat = diff_avg_earnings / se_diff_avg_earnings
t_stat
```

```python
conf_int = (diff_avg_earnings - 1.96 * se_diff_avg_earnings,
           diff_avg_earnings + 1.96 * se_diff_avg_earnings)
conf_int
```

<div class="ex">
    
### Exercise 4c
*   The lazy way: use the inbuilt ``tconfint_diff`` function of the `CompareMeans` class. [1 mark]

</div>

```python
wabove = df_clean[df_clean.wabove == 1]
wbelow = df_clean[df_clean.wabove == 0]

ds1 = sms.DescrStatsW(wabove.earnings)
ds2 = sms.DescrStatsW(wbelow.earnings)
sms.CompareMeans(ds1, ds2).tconfint_diff(usevar="unequal")
```

<div class="ex">
    
### Exercise 4d
Use both ways of calculating and testing the earnings gap to answer this question: Can you distinguish the earnings difference statistically from zero?  Write your answer in a text cell.

From the above tests, we observe:
- The test statistic t = 3.8077 > 1.96  (or -3.8077 < -1.96, depending on which direction we use to take the difference in means)
- The 95% confidence interval does not include the value 0

We conclude that we can distinguish the earnings difference statistically from zero (given a two-sided hypothesis test with a 5% significance level).

## Exercise 5
Run a regression of ``earnings`` on ``weightkg``. Create a scatter plot and include the estimated population regression function. What is the direction of the statistical association between earnings and weight? Is it significant?  [1 mark]

</div>

```python
formula = 'earnings ~ weightkg'
model1 = smf.ols(formula, data=df_clean, missing='drop')
reg1 = model1.fit()
reg1.summary()
```

```python
fig, ax = plt.subplots()
ax.scatter(df_clean.weightkg, df_clean.earnings, marker=".")
ax.set_title('Earnings vs weights')
ax.set_xlabel('Weight (kg)')
ax.set_ylabel('Annual earnings ($)')

x = np.linspace(35, 190, 10)
prf = reg1.params[0] + reg1.params[1]*x
ax.plot(x, prf, color = 'red', linewidth=3)

plt.show()
```

The estimated PRF implies a slightly positive statistical association between earnings and weight.  The point estimate for the slope coefficient of the PRF is 28.07; that is, for each kilogram increase in weight, expected earnings increase by $28.07.

We obtain the following information from the regression output and confidence intervals for the coefficient estimates:

- The test statistic for the slope coefficient estimate is 2.397 > 1.96
- The confidence interval for the slope coefficient is [5.113136, 51.02903], which does not include zero
- The p-value for the slope coefficient estimate is 0.01656 < 0.05

Therefore, we can say that there is statistical evidence to reject the null hypothesis that the true slope coefficient of the PRF is equal to zero (given a two-sided hypothesis test with a 5% significance level).  That is, the association between earnings and weight is statistically significant.

## Exercise 7
Does weight have an effect on earnings? Provide a brief yet thoughtful discussion! [2 marks]
    
<div>

As mentioned in Exercise 6, the estimate for the slope coefficient of the PRF is 28.0711.  The interpretation of this estimate is that for each kilogram increase in weight, expected earnings increase by around $28.07.  This can be seen on the scatterplot - the estimated PRF has a very slightly positive slope.

We also noted in Exercise 6 that sufficient statistical evidence exists to conclude that the true slope coefficient of the PRF is not equal to zero (given a two-sided alternative hypothesis with a 5% significance level). However, the slope coefficient estimate does not imply that weight has a 'large' effect on earnings, in terms of economic importance.  For example, a ten-kilogram difference in weight between two individuals would result in an expected difference in annual earnings of only $280.71.

Furthermore, visual inspection of the scatterplot also shows that the observations in the sample still vary widely around the estimated PRF.  That is, the OLS regression line does not fit the data well. To corroborate this observation, we also note that the adjusted R-squared value from the regression is <0.001 (a very low value), indicating that only a very small fraction of the sample variance in earnings is explained by weight.

It is also important to note that this regression may be subject to omitted variable bias - it is well-documented that males, on average, have greater weights than females, and therefore this regression may also be capturing, at least to some extent, the effect of gender on earnings.  Further analysis, for example by separating the data into two sub-samples according to gender and re-performing the regression on those sub-samples, is warranted to better investigate the true causal effect of weight on earnings.
