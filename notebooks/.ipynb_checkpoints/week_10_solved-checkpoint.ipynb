## Imports and loading data

```python
import matplotlib.pyplot as plt
import pandas as pd
import statsmodels.formula.api as smf
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.tsa.ar_model import AutoReg
import numpy as np
```

```python
# from google.colab import drive
# drive.mount('/content/drive')
# df = pd.read_csv('drive/MyDrive/EMET2007/datasets/cpi_aus_2025.csv')
```

```python
df = pd.read_csv('../datasets/cpi_aus_2025.csv')
```

Run this cell once you've loaded the data:

```python
df['date'] = pd.to_datetime(df['time'], format='%b-%Y')
df.index = pd.DatetimeIndex(df.date, name='quarter').to_period('Q')
```

## Exercise 1

Create two new time series variables in the dataset: ``cpi`` and ``infl``. The ``infl`` time series is created as annualized log-differences. Use the `head` and `tail` functions to look at the first two years of both time series, and again to look at the last two years.


```python
df['logcpi'] = np.log(df.cpi)
df['infl'] = 400 * df.logcpi.diff()
df.head(8)
```

```python
df.tail(8)
```

   
## Exercise 2

Plot both time series. Look what happened to inflation over time!


```python
fig, axs = plt.subplots(2, 1, figsize=(8,7))

axs[0].plot(df.date, df.cpi)
axs[0].set_ylabel('CPI')
axs[0].set_title('CPI (Australia)')
axs[1].plot(df.date, df.infl)
axs[1].set_ylabel('Inflation (annualised, %)')
axs[1].set_title('Inflation (Australia)')
fig.supxlabel('Time')
plt.show()
```

   
## Exercise 3

Create two new time series: ``l1cpi`` and ``l1infl`` which are lagged versions of order 1 of ``cpi`` and ``infl``.  Use the ``head`` command to look at the first two years of ``cpi``, ``l1cpi``, ``infl``, ``l1infl``, and the `tail` command to look at the last two years.


```python
df['l1cpi'] = df.cpi.shift(1)
df['l1infl'] = df.infl.shift(1)
df[['date', 'cpi', 'l1cpi', 'infl', 'l1infl']].head(8)
```

```python
df[['date', 'cpi', 'l1cpi', 'infl', 'l1infl']].tail(8)
```


## Exercise 4 

What is the correlation between ``cpi`` and ``l1cpi``? Between ``infl`` and ``l1infl``? Explain the meaning of serial- or auto-correlation in this context.


```python
df[['cpi', 'l1cpi']].corr()
```

```python
df[['infl', 'l1infl']].corr()
```

**The meaning of serial- or auto-correlation in this context:** the 
correlation of a time series with its own lagged values (here we have computed the 1st sample autocorrelations for cpi and infl).

## Exercise 5 

Use the ``plot_acf`` command we imported at the start to automatically generate a plot of autocorrelations for ``cpi`` and ``infl``. 

```python
plot_acf(df.cpi)
plt.show()
```

```python
plot_acf(df.infl, missing='drop')
plt.show()
```


## Exercise 6

Run a first order autoregression for ``infl``.



```python
# 'Manually', using the same method we've used through the semester
reg1 = smf.ols('infl ~ l1infl', data=df, 
               missing='drop').fit(use_t=False)
reg1.summary()
```

```python
?AutoReg
```

```python
# Or using the built-in class
ar1 = AutoReg(df.infl, lags=1, missing='drop').fit(use_t=False)
ar1.summary()
```

**Note:** 
- 407 obs on original time series, which seems to be what is reported in the results, but the regression that runs under the hood uses the correct number
- SE differences are based on the df adjustments they use, but don't worry about this


## Exercise 7
    
The data set contains data up to (and including) 2025:Q1. This tutorial is taking place during 2025:Q2. We can use the estimated AR(1) model now to create the following:

* a prediction for 2025:Q1
* a forecast for 2025:Q2

For both the prediction and the forecast, create them in two different ways: manually and alternatively using the ``predict`` function.



```python
reg1.params
```

```python
print(f'Prediction for 2025:Q1 = {reg1.params['Intercept'] + reg1.params['l1infl'] * df.infl['2024Q4']}')
print(f'Forecast for 2025:Q2 = {reg1.params['Intercept'] + reg1.params['l1infl'] * df.infl['2025Q1']}')
```

```python
newdata = {'l1infl' : [df.infl['2024Q4'], df.infl['2025Q1']]}
reg1.predict(newdata)
# Prediction for 2025:Q1, and forecast for 2025:Q2
```

```python
estimates = ar1.predict(start='2025Q1', end='2025Q2')
estimates
```

**Notes:**
- Recall: do not use heteroskedasticity-robust option here as we only care about prediction / forecasting here.  Confidence intervals etc. (for which we need the standard errors) are for higher-level time series applications
