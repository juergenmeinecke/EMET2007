## Exercise 4 

Now your tutor will show you one way of constructing boxplots in `Python`.  Construct a boxplot of female life expectancy. How do you interpret the resulting graph?

```python
fig, ax = plt.subplots()

ax.boxplot(df.lifeExpecFem.dropna(), labels=['Females'])
ax.set_ylabel('Life expectancy at birth (years)')
ax.set_title('Boxplot of female life expectancy (WDI)')
ax.grid()
plt.show()
```

**Interpretation:**
- Min approx. 53
- Q1 approx. 69
- Median approx. 77
- Q3 approx. 81
- Max approx. 88
- The middle 50% of female life expectancy is between approx. 69 and 80.5
- No values that are considered outliers by the boxplot function (ie values outside of 1.5*IQR)
- Distribution seems left-skewed (long left tail)

We can check these figures:

```python
df.lifeExpecFem.describe()
```

## Exercise 6
Can you summarize `lifeExpecFem` in a histogram instead? Both box plots and histograms are visual ways of describing **distributions** of variables.

Use this opportunity to get familiar with the built-in help function. In a new cell in your Jupyter notebook, type ``?plt.hist`` and run the cell.  Give your histogram a title, label its axes, and explore the ``density`` option (switching between absolute and relative frequencies).

```python
?plt.hist
```

```python
fig, ax = plt.subplots()

ax.hist(df.lifeExpecFem, ec='black', bins=np.arange(45,100,5))
ax.set_xlabel('Life expectancy at birth (years)')
ax.set_ylabel('Number of countries')
ax.set_title('Histogram of female life expectancy (WDI)')
plt.show()
```

**Tutor's note**: Warning with using `np.arange()` - much like the `range()` function in Python - the upper limit is excluded. So saying `bins=np.arange(45,100,5)` is like saying 'let my first bin start at 45, then proceed in intervals of 5 until 95'.  If you were to set `bins=np.arange(50,90)`, which would seem reasonable from the boxplot, the 85-90 bin will be **excluded** from the graph.

Because of this, it seems best to teach students to define their bin ranges such that they leave some empty space on either side of their histogram.

```python
fig, ax = plt.subplots()
ax.hist(df.lifeExpecFem, ec='black', 
        density=True, bins=np.arange(45,100,5))
ax.set_xlabel('Life expectancy at birth (years)')
ax.set_ylabel('Density')
ax.set_title('Histogram of female life expectancy (WDI)')
plt.show()
```

**Tutor's note**: The densities don't sum to 1 vertically, but the *areas* of the bars should sum to 1.

* one categorical and one numerical variable

## Exercise 8
    
The association between two categorical variables is typically presented in a contingency table. Use ``pandas``' ``crosstab`` function to study the association between region and income group.  Also explore the `normalize` argument of this function.

```python
?pd.crosstab
```

```python
pd.crosstab(df.region, df.incomeGroup, margins = True)
```

```python
pd.crosstab(df.region, df.incomeGroup, margins = True, normalize = True)
# add "* 100" on the end to get percentages instead of proportions
```

## Exercise 9
    
We can group the observations in the dataset by a categorical variable using the `groupby` function.  We can then use this grouping to explore the association between a categorical and a numerical variable.  Explore the association between per capita GDP and income group.

```python
df.groupby('incomeGroup').GDPpc.describe()
```

## Exercise 10
The association between two numerical variables can be condensed in two popular measures: covariance and correlation. Remember that these measures only summarize the **linear** association between two variables. Calculate the covariance and correlation between per capita GDP and female life expectancy at birth. Do they seem strongly associated?

```python
df[['GDPpc', 'lifeExpecFem']].cov()
```

```python
df[['GDPpc', 'lifeExpecFem']].corr()
```

**Interpretation:** A correlation coefficient of 0.625464 implies a moderate positive correlation between the two variables.

## Exercise 11
Now generate a scatter plot of these two variables. Does the relationship between per capita GDP and female life expectancy at birth appear linear? Do they seem strongly associated?

```python
fig, ax = plt.subplots()

ax.plot(df.GDPpc, df.lifeExpecFem, '.', c='darkgreen')
ax.set_xlabel('GDP per capita ($)')
ax.set_ylabel('Life expectancy (years)')
ax.set_title('Per capita GDP vs female life expectancy at birth')
plt.show()
```

**Interpretation:** Association seems strong, but relationship does not appear linear.
