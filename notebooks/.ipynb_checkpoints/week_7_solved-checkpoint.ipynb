## Exercise 1

Run a regression of ``growth`` on ``tradeShare`` (make sure to always exclude Malta in all your estimations).  Test the null hypothesis that $\beta_1 = 0$ using three alternative ways: 
via the t-statistic, p-value, and the confidence interval.  Do this under alternative model assumptions:

*   OLS Assumption 4a: homoskedasticity
*   OLS Assumption 4b: heteroskedasticity

### Homoskedasticity

```python
df_clean = df[df.tradeShare < 1.5].copy()
formula = 'growth ~ tradeShare'
reg1 = smf.ols(formula, data=df_clean, missing='drop').fit(use_t=False)
reg1.summary()
```

### Heteroskedasticity

```python
reg2 = smf.ols(formula, data=df_clean, missing='drop').fit(cov_type='HC1', use_t=False)
reg2.summary()
```

What do these four variables capture? Why would it be useful to include them in a multiple regression model?

* rgdp60: The starting value of a country's real GDP - may be a predictor of the country's future growth (e.g. convergence hypothesis)
* yearsSchool: The level of education of a country's citizens - may be a predictor of the country's capacity for growth (productivity)

* assassinations: Average number of political assassinations per year over the period

You now are going to run a **multiple** regression including four additional regressors to explain variation in GDP growth.

## Exercise 3

### Exercise 3a
Run a regression of ``growth`` on ``tradeShare``, ``rgdp60``, ``yearsSchool``, ``revCoups``, and ``assassinations``. 

```python
formula = 'growth ~ tradeShare + rgdp60 + yearsSchool + revCoups + assassinations'
reg3 = smf.ols(formula, data=df_clean, missing='drop').fit(cov_type='HC1', use_t=False)
reg3.summary()
```

### Exercise 3b
Interpret the coefficient on ``tradeShare``. Is it statistically significant? Compare it to the estimate you obtained under the simple regression model.

The coefficient on tradeShare is not statistically significant:
- t-statistic = 1.520 < 1.96
- p-value = 0.128 > 0.05
- confidence interval includes the value 0

The coefficient estimate on tradeshare is 1.3408..., compared to 1.6809... on the simple regression model.  As this coefficient estimate has decreased substantially with the inclusion of more regressors, the simple regression was likely to have been subject to omitted variable bias (i.e. in the simple regression, the coefficient estimate for tradeshare was also capturing the effects of other variables).

Aside: Under homoskedasticity, rev_coups is not significant at the 5% level, but under heteroskedasticity, it becomes significant at the 5% level

## Exercise 4
What is the value of the coefficient on ``revCoups``? Interpret the value of this coefficient. Is it large or small in a real-world sense?

**Interpretation:** The value of the coefficient is -2.1504, which implies that for every additional 35 revolutions/insurrections/coup d'etats in the country over the specified 35-year period, average GDP growth is estimated to fall by 2.1504%.

In a real-world sense, that large a difference in GDP growth is extremely significant.  However, it would seem extremely unlikely for most countries to experience this level of civil unrest, with none of the countries in the dataframe having experienced more than 35 in the period (although Bolivia and Argentina come close).

## Exercise 5
Use the regression to predict the average annual growth rate for a country that has average values for all regressors.

```python
newdata = {'tradeShare' : df_clean.tradeShare.mean(),
           'rgdp60' : df_clean.rgdp60.mean(),
           'yearsSchool' : df_clean.yearsSchool.mean(),
           'revCoups' : df_clean.revCoups.mean(),
           'assassinations' : df_clean.assassinations.mean()
          }
reg3.predict(newdata)
```

**Note:** this is equal to the mean of the `growth` variable (predictably).

```python
df_clean.growth.mean()
```

Repeat the previous exercise but now assume that the country's value for ``tradeshare`` is one standard deviation above the mean.

```python
newdata = {'tradeShare' : df_clean.tradeShare.mean() + df_clean.tradeShare.std(),
           'rgdp60' : df_clean.rgdp60.mean(),
           'yearsSchool' : df_clean.yearsSchool.mean(),
           'revCoups' : df_clean.revCoups.mean(),
           'assassinations' : df_clean.assassinations.mean()
          }
reg3.predict(newdata)
```

## Exercise 7
Test whether, taken as a group, ``rgdp60``, ``yearsSchool``, ``revCoups``,  and ``assassinations`` can be omitted from the regression. Recall that the null hypothesis here is

$$ \beta_2 = \beta _3 = \beta_4 = \beta_5 = 0 $$

Where the underlying model is

$$ \texttt{growth} = \beta_0 + \beta_1 \cdot \texttt{tradeShare} + \beta_2  \cdot  \texttt{rgdp60} + \cdots + \beta_5  \cdot \texttt{assassinations} + u  $$

What is the p-value of the *F*-statistic? Based on the p-value, do you reject the null hypothesis?

```python
ftest = reg3.f_test('rgdp60 = yearsSchool = revCoups = assassinations = 0')
ftest
```

**Interpretation:** The p-value of the F-statistic is 0.00002649 < 0.05 => we reject the null hypothesis.
