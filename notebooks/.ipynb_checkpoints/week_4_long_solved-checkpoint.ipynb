## Exercise 4 

Add a new (categorical) variable ``habove`` to your data frame. The definition of ``habove`` is as follows: it equals 1 if ``height`` exceeds 67 inches (it is **above** the median height in the original sample), and zero otherwise.

(This makes ``habove`` a so-called *dummy* variable which splits the sample in two.)

For the two sub-samples that are created by ``habove``, collect the following information for ``earnings``:


|  Information | ``habove == 0`` | ``habove == 1`` |
|:--|:--|:--|
| Sample size *n* | | |
| Average | | | 
| Standard deviation | | |
| Standard error | | |


```python
df['habove'] = df.height.apply(lambda x: 1 if x > 67 else 0)
```

```python
df.head()  # check to see it was correctly created
```

```python
df_n = df.groupby('habove').earnings.count().rename('n')
df_mean = df.groupby('habove').earnings.mean().rename('average')
df_sd = df.groupby('habove').earnings.std().rename('sd')
df_se = df.groupby('habove').earnings.sem().rename('se')
```

```python
results_df = pd.concat((df_n, df_mean, df_sd, df_se), axis=1)
results_df
```

## Exercise 5 

Can you distinguish the earnings difference statistically from zero? Construct a 95% confidence interval for the earnings difference between the two groups. Do it in two different ways:
    
### Exercise 5a
The hard way: Implement the appropriate statistical test by programming in ``Python``. The information in the table above will be useful.
    
Here is a gentle reminder (see section 3.4 in Stock and Watson): Under the null hypothesis of a zero difference in means,
    
</div>
$$ t := \frac{(\overline{Y}_1 - \overline{Y}_0)}{\text{se} (\overline{Y}_1 - \overline{Y}_0)} \sim N(0,1)$$
$$ \text{se} (\overline{Y}_1 - \overline{Y}_0) := \sqrt{s_0^2/n_0 + s_1^2/n_1}\\ := \sqrt{\text{se} (\overline{Y}_1)^2 + \text{se} (\overline{Y}_0)^2} $$


```python
diff_avg_earnings = results_df.average[1] - results_df.average[0]
# Important note: 1 and 0 here are the values of habove, not row indexes

se_diff_avg_earnings = np.sqrt(results_df.se[0]**2 + results_df.se[1]**2)
t_stat = diff_avg_earnings / se_diff_avg_earnings
t_stat
```

```python
conf_int = (diff_avg_earnings - 1.96 * se_diff_avg_earnings,
           diff_avg_earnings + 1.96 * se_diff_avg_earnings)
conf_int
```

## Exercise 7
    
Run a regression of ``earnings`` on ``height``.

### Exercise 7a
Inspect the standard ``Python`` regression summary output and make sense of the results.

```python
formula = 'earnings ~ height'
model1 = smf.ols(formula, data=df, missing='drop')
reg1 = model1.fit(use_t=False)
reg1.summary()
```

**Interpretation:**
- If you have 0 height you are expected to have \$-512.73 earnings annually (no sensible interpretation here)
- For each inch you grow it increases by \$707.67
- Note: confidence intervals for the regression coefficients are included in the regression output

### Exercise 7b
Add your estimated PRF to the scatter plot.

```python
fig, ax = plt.subplots()
ax.scatter(df.height, df.earnings, marker=".")
ax.set_title('Earnings vs heights')
ax.set_xlabel('Height without shoes (in)')
ax.set_ylabel('Annual labour earnings ($)')

x = np.linspace(48, 84, 10)
prf = reg1.params[0] + reg1.params[1]*x
ax.plot(x, prf, color = 'red', linewidth=3)
plt.show()
```

### Exercise 7c
Use your estimated PRF to create `predictions` for earnings of a worker who is 65 inches tall / 67 inches tall / 70 inches tall.
    

```python
data = {'height' : [65, 67, 70]}
reg1.predict(data)
```
