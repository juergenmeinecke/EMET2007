### `courseEval`

From the boxplot:

- The median course_eval (average overall evaluation score for a course) sits at around 4.
- There are a small number of outlier values on the left side of the distribution.
- The maximum average evaluation score for a course in the dataset was 5 (the highest possible score), and these values are not considered outliers (recalling that for the boxplot function in Python, the default definition of an outlier is one that is more than 1.5 * IQR away from the box).
- Despite course evaluation scores theoretically being able to range from 1 to 5, the minimum observed value in the dataset was 2.1.


From the histograms:
- It is more easily observed that course_eval has a left-tailed distribution.
- The vast majority of course_eval scores lie between 3 and 5.
- The most common range for course_eval is between 3.8 and 4.0 (based on the second of the three histograms).


Interpretations:
- It was more common, on average, for courses to receive scores towards the 'excellent' (the definition of a 5 score, according to the dataset 
  description) end of the spectrum of scores, rather than towards the 
  'very unsatisfactory' (the definition of a 1 score, according to the dataset description) end of the spectrum of scores.
- Even for courses that received low average scores, there were no instances observed where students (near-)unanimously voted a course to be 'very unsatisfactory' (evidenced by the fact that the minimum observed value for course_eval was 2.1).
- Some courses were (near-)unanimously voted 'excellent' (average score of 5) by the students that undertook them.  However, class sizes were not provided for these courses - or indeed for any particular course in the dataset.  It would be interesting to see if there was any association between class size and average course evaluation score as a matter for further study.
- Overall, the average course evaluation score was close to 4, indicating that, on average, students found these particular courses 'very good' during the period in which the data were collected.

## Exercise 2

Run a simple regression of ``courseEval`` on the professor's `beauty`.  Create a scatter plot and add the estimated PRF.  Does there appear to be an association between the variables?
    
</div>

```python
reg1 = smf.ols('courseEval ~ beauty', 
               data=df, 
               missing='drop').fit(cov_type='HC1', use_t=False)
reg1.summary()
```

```python
fig, ax = plt.subplots()
ax.scatter(df.beauty, df.courseEval, marker=".")
ax.set_title('courseEval vs beauty')
ax.set_xlabel('Beauty rating')
ax.set_ylabel('Course evaluation score')

x = np.linspace(-1.6, 2, 10)
prf = reg1.params[0] + reg1.params[1]*x
ax.plot(x, prf, color = 'red', linewidth=2)

plt.show()
```

There does appear to be an association between the variables.  For the slope coefficient estimate:
- The t-statistic is 4.1153 > 1.96
- The p-value is 0.00004581 < 0.05
- The 95% confidence interval does not include 0

Therefore we conclude that the slope coefficient of beauty is not equal
to zero (given a two-sided hypothesis test with a 5% significance level).

The slope coefficient estimate itself implies that for a 1-unit increase in the beauty variable, course_eval is predicted to increase by around 0.133001.

Graphically, the estimated PRF shows a slight positive association between beauty and course_eval.

However, we note that these results are not necessarily suggestive of any causal effect of beauty on course_eval.

## Exercise 5
Run the same multiple regression as in part (4) but exclude the variable ``female``. What is the coefficient estimate on ``beauty`` equal to? How has it changed? What does this tell you about omitted variables bias?  Give a brief yet thoughtful discussion!


</div>

```python
formula3 = 'courseEval ~ beauty + intro + oneCredit + minority + nnEnglish'
reg3 = smf.ols(formula3, 
               data=df, 
               missing='drop').fit(cov_type='HC1')
reg3.summary()
```

The coefficient estimate on beauty is 0.1526 (down from 0.1656, a decrease of 0.013). This implies that in the regression with the variable female omitted, the estimated effect of beauty also indirectly captures the effect of being female and the regression is therefore prone to omitted variables bias. 

More precisely, in the regression without the female variable, the error term incorporates the female variable.  We suspect that being female is at least somewhat correlated with beauty - which we can confirm for the sample data:

```python
df[['beauty', 'female']].corr()
```

Therefore, for this regression without the female variable, OLS Assumption 1 is violated: the error term is correlated with at least one of the regressors and so cannot be CMI of the regressors.

We also suspect that being female may be a determinant of course evaluations (that is, the true population coefficient for the variable female is not equal to zero) - and the results from Exercise 4 certainly suggest this. Therefore, the two conditions for omitted variable bias are met.

We can further discuss the direction of the bias in this example.  We know that the direction of the bias in the coefficient estimate for a certain regressor depends on the signs of both:
- the covariance between that regressor and the omitted variable, and
- the true slope coefficient on the omitted variable.

Here we have observed a negative bias in the coefficient estimate for beauty.This makes sense as we suspect a positive covariance between the variables beauty and female (based on the sample correlation calculated above), and we also suspect that the true effect of being female on course evaluations is negative (based on the coefficient estimates and confidence intervals calculated in Exercise 4).  

When only two regressors are involved (one included and one omitted), we know that:

$$ E [ \widehat{\beta}_1 | X_{1i}, X_{2i} ] = \beta_1 + \beta_2 \frac{Cov(X_{1i}, X_{2i})}{Var (X_{1i})} $$

where the bias is given by the second term.  While we cannot apply this exact formula to this example, conceptually we still suspect that a negative $ \beta_2 $ and a positive covariance should yield a negative bias on the coefficient estimate for beauty - which, indeed, we have observed above.

As a final note, we observe that the coefficient estimate on beauty was not the only coefficient estimate to change when the female variable was omitted from the regression.  Therefore, the above discussion of omitted variables bias likely applies in a similar fashion to many of the other regressors in this multiple regression.

## Exercise 6
    
Modify the model so that the effect of ``beauty`` on ``courseEval`` is different for men and women.  (Hint: the intended modification does NOT consist of splitting the sample and running separate regressions.) Is the male/female difference in the effect of ``beauty`` statistically significant? 

</div>

```python
formula = 'courseEval ~ beauty*female + intro + oneCredit + minority + nnEnglish'
reg4 = smf.ols(formula, 
               data=df, 
               missing='drop').fit(cov_type='HC1', use_t=False)
reg4.summary()
```

For the variable (beauty * female) - i.e. the estimated difference between being male and being female on the effect of beauty:
- The p-value is 0.026 < 0.05.  
- The t-statistic is -2.221 < -1.96.
- The confidence interval does not include the value zero.

Therefore there is sufficient evidence to conclude that the interaction term is statistically significant at the 5% significance level.  That is, the male/female difference in the effect of beauty is statistically significant.

## Exercise 7

### Exercise 7a
Professor White has cosmetic surgery that increases his beauty index from one standard deviation below the average to one standard deviation above the average. What is his value of ``beauty`` before the surgery? After the surgery? 

<div>

Before the surgery, Professor White's value for `beauty` is:

```python
df.beauty.mean() - df.beauty.std()
```

After the surgery, Professor White's value for `beauty` is:

```python
df.beauty.mean() + df.beauty.std()
```

<div class="ex">

### Exercise 7b
By how much does his predicted course evaluation increase? 
    
<div>

Professor White's predicted course evaluation increases by:

$$ \Delta X_1 \cdot \beta_1  = 2 \sigma_1 \cdot \beta_1$$

where $X_1$ is `beauty`, $ \sigma_1 $ is the standard deviation of `beauty`, and $ \beta_1 $ is the coefficient on `beauty` in the multiple regression.

```python
2 * df.beauty.std() * reg4.params[1]
```

<div class="ex">
 
### Exercise 7c
Construct a 95% confidence interval for the increase in his course evaluation. 
    
<div>

Confidence interval for the increase in courseEval, given a **unit** change in beauty:

```python
reg4.conf_int().loc['beauty']
```

Confidence interval for the increase in courseEval, given Professor White's change in beauty:

```python
reg4.conf_int().loc['beauty'] * 2 * df.beauty.std()
```

Therefore, the 95% confidence interval for the increase in Professor White's course evaluation is [0.216356, 0.511785].

### Exercise 8b
By how much does her predicted course evaluation increase? 
    
<div>

Professor Robinson's predicted course evaluation increases by:

$$ \Delta X_1 \cdot (\beta_1 + \beta_3)  = 2 \sigma_1 \cdot (\beta_1 + \beta_3)$$

where:
- $X_1$ is `beauty`, 
- $ \sigma_1 $ is the standard deviation of `beauty`, 
- $ \beta_1 $ is the coefficient on `beauty` in the multiple regression, and
- $ \beta_3 $ is the coefficient on the interaction term in the multiple regression.

```python
2 * df.beauty.std() * (reg4.params[1] + reg4.params[3])
```

### Exercise 8c
Construct a 95% confidence interval for the increase in her course evaluation. 
    
<div>

```python
df['male'] = 1 - df.female
formula_male = 'courseEval ~ beauty*male + intro + oneCredit + minority + nnEnglish'
male_reg = smf.ols(formula_male, 
               data=df, 
               missing='drop').fit(cov_type='HC1')
male_reg.summary()
```

## Exercise 9
    
### Exercise 9a
Add a linear and a quadratic ``age`` term to the regression. 

</div>
    

```python
formula = 'courseEval ~ beauty*female + intro + oneCredit + minority + nnEnglish + age + I(age**2)'
reg5 = smf.ols(formula, 
               data=df, 
               missing='drop').fit(cov_type='HC1')
reg5.summary()
```

<div class="ex">
    
### Exercise 9b
Is there evidence that ``age`` has a nonlinear effect on `courseEval`? Is there evidence that ``age`` has any effect on ``courseEval``? 
    
<div>

The quadratic term for age is not significant: 
- p-value = 0.5385 > 0.05;
- t-stat = -0.615 > -1.96;
- 95% confidence interval contains zero.

This suggests that age does not have a nonlinear effect on course_eval.

We also wish to determine if there is any evidence that age has any effect on course_eval.  We therefore run an F test to test the joint hypothesis that both of the two age terms are equal to zero, against the alternative hypothesis that either one is nonzero:

```python
ftest = reg5.f_test('age = I(age ** 2) = 0')
print(ftest)
```

The p-value for the F test is 0.7443 > 0.05, implying that we do not reject the null hypothesis that both of the age terms are equal to zero.  That is, we conclude that there is no evidence that age has any effect (linear or nonlinear) on course_eval.
