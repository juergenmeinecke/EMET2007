## Exercise 1

### Exercise 1a
Run a regression of ``earnings`` on ``heightcm``.


```python
df['heightcm'] = df.height * 2.54
```

```python
formula = 'earnings ~ heightcm'
model1 = smf.ols(formula, data=df, missing='drop')
reg1 = model1.fit()
reg1.summary()
```

### Exercise 1b
How do intercept and slope estimate change compared to last week's exercise (in which you used heights measured in inches)?

**Observation:** 
- The intercept estimate is the same, which makes sense as 0in = 0cm
- The slope estimate for heightcm is equal to (slope estimate for height) / 2.54

## Exercise 3
Last week you studied the bivariate association between ``earnings`` and ``height``. You found that both are positively associated. Recall that the average annual earnings gap  between the two groups ``habove==0`` and ``habove==1`` was about \$5,500. 

### Exercise 3a
We want to explore if this gap should be taken at face value. That is, does height really **cause** earnings to increase by that amount? To do so, study the bivariate associations between 
- ``earnings`` and ``male``, and
- ``heightcm`` and ``male``.
    
(Use descriptive statistics to study these associations.)


```python
earnings_n = df.groupby('male').earnings.count().rename('n')
earnings_mean = df.groupby('male').earnings.mean().rename('average')
earnings_sd = df.groupby('male').earnings.std().rename('sd')
earnings_se = df.groupby('male').earnings.sem().rename('se')
```

```python
male_earnings = pd.concat((earnings_n, earnings_mean, 
                           earnings_sd, earnings_se), 
                          axis=1)
male_earnings
```

```python
heightcm_n = df.groupby('male').heightcm.count().rename('n')
heightcm_mean = df.groupby('male').heightcm.mean().rename('average')
heightcm_sd = df.groupby('male').heightcm.std().rename('sd')
heightcm_se = df.groupby('male').heightcm.sem().rename('se')
```

```python
male_heightcm = pd.concat((heightcm_n, heightcm_mean, 
                           heightcm_sd, heightcm_se), 
                          axis=1)
male_heightcm
```

### Exercise 3b
What do you learn here that helps you shape a more informed opinion about the causal effect of height on earnings? 

**Interpretation**

Last week:
- our regression suggested that taller people earn more on average

This week:
- men are 15cm taller on average
- men earn about $2,800 more on average

Some of the association between earnings and heights may be spurious: while it may capture to some degree the pure effect of height on earnings, it also partially captures the effect of being male on earnings.

## Exercise 4

### Exercise 4a
Run a regression of ``earnings`` on ``heightcm`` in two sub-samples:
* females
* males


```python
females = df[df.male == 0]
males = df[df.male == 1]
```

```python
formula = 'earnings ~ heightcm'
model1 = smf.ols(formula, data=females, missing='drop')
reg1 = model1.fit()
reg1.summary()
```

```python
model2 = smf.ols(formula, data=males, missing='drop')
reg2 = model2.fit()
reg2.summary()
```

### Exercise 4b

How does this address some of the lessons learned in the preceding exercise?

- This is one way of addressing the clear OVB problem that we noted in 3b
- We also observe that height has more of an influence on earnings for males than for females
