# EMET2007 Week 6

In this exercise, you will investigate, using OLS estimation, the research question: 

**Do countries with open economies have higher growth rates?** 

(What would be your null hypothesis?)

To answer this question, you will use the Growth data set which is a cross-country data set that contains information on growth rates and trade openness for 65 countries for the years 1960 to 1995; you can find its description on the course website.

The two main variables that we will associate with each other today are:
* ``growth``: average annual real GDP growth rate between 1960 and 1995

* ``tradeShare``: average annual value of (Imports+Exports)/GDP between 1960 and 1995
  (as a proxy for economic openness)

## Imports and loading data

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
```

```python
# from google.colab import drive
# drive.mount('/content/drive')
# df = pd.read_csv('drive/MyDrive/EMET2007/datasets/growth.csv')
```

```python
df = pd.read_csv('../datasets/growth.csv')
```

## Exercise 1

Look at boxplots and histograms of ``growth`` and ``tradeShare``. Do you detect anything of interest? (For example, one country is conspicuous with regard to its economic openness. Which country is it? What is going on here?)



```python
# Can do as subplots, but separate plots work fine too
fig, axs = plt.subplots(1, 2, figsize=(12,5))

axs[0].boxplot(df.growth, vert=False)
fig.suptitle('Average annual growth in real GDP, 1960-1995', fontsize=15)
axs[0].set_xlabel('Average annual GDP growth (%)')
axs[1].hist(df.growth, ec='black', bins=np.arange(-4, 10, 1))
axs[1].set_xlabel('Average annual GDP growth (%)')
axs[1].set_ylabel('Frequency')
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(12,5))
axs[0].boxplot(df.tradeShare, vert=False)
fig.suptitle('Average annual value of (Imports + Exports) / GDP, 1960-1995', fontsize=15)
axs[0].set_xlabel('Average annual value for (M + X) / GDP')
axs[1].hist(df.tradeShare, ec='black', bins=np.arange(0, 2.1, 0.05))
axs[1].set_xlabel('Average annual value for (M + X) / GDP')
axs[1].set_ylabel('Frequency')
plt.show()
```

```python
df[df.tradeShare > 1.5]  # the conspicuously open country
```

**Interpretation:** The island nation of Malta has a conspiciously high value for average annual trade share, almost 2.  This implies that, on average over the period in question, imports and exports together accounted for around double Malta's domestic production.  This makes sense as Malta is a small island nation that is well-positioned to (and probably highly reliant on) trade with its neighbours in the Mediterranean.

## Exercise 2
Create a scatter plot between the average annual growth rate and average trade share.  Is the graph suggestive of an association between the two? Which direction?

```python
fix, ax = plt.subplots()

ax.scatter(df.tradeShare, df.growth, marker='.')
ax.set_title('growth vs tradeShare', fontsize=15)
ax.set_xlabel('Average annual value for (M + X) / GDP')
ax.set_ylabel('Average annual GDP growth (%)')
plt.show()
```

**Interpretation:** Unclear about sign of association, perhaps slightly positive?

## Exercise 3
Run a regression of ``growth`` on ``tradeShare``. What are the estimates for the intercept and the slope? 

```python
formula = 'growth ~ tradeShare'
model1 = smf.ols(formula, data=df, missing='drop')
reg1 = model1.fit(use_t=False)
reg1.summary()
```

Estimates for intercept and slope are:

```python
reg1.params
```

## Exercise 4
Test the null hypothesis that $\beta_1 = 0$ using three alternative ways: via the t-statistic, p-value, and the confidence interval.

Reading from regression output above:
- t-statistic = 2.982 > 1.96
- p-value = 0.004 < 0.05
- confidence interval does not include zero

Therefore we conclude that there is sufficient evidence at the 95% significance level, given a two-tailed alternative hypothesis, that the population slope coefficient is not equal to zero.

## Exercise 5
Use the regression to predict the growth rate for a country with a trade share of 0.5 and for another with trade share of 1.0.

```python
newdata = {'tradeShare' : [0.5, 1.0]}
reg1.predict(newdata)
```

## Exercise 6

### Exercise 6a
Run a regression of ``growth`` on ``tradeShare``, this time excluding the outlier from above.  


```python
df_clean = df[df.tradeShare < 1.5].copy()
reg2 = smf.ols(formula, data=df_clean, missing='drop').fit(use_t=False)
reg2.summary()
```

**Note:** We no longer reject the null hypothesis of a zero slope at the 5% level of significance.


### Exercise 6b
Use this new regression to predict the growth rate for a country with a trade share of 0.5 and for another with trade share of 1.0.


```python
reg2.predict(newdata)
```

## Exercise 7
Plot the two estimated PRFs. 

```python
fix, ax = plt.subplots()

ax.scatter(df.tradeShare, df.growth, marker='.', color='black')
ax.set_title('growth vs tradeShare', fontsize=15)
ax.set_xlabel('Average annual value for (M + X) / GDP')
ax.set_ylabel('Average annual GDP growth (%)')

x = np.linspace(0.1, 2, 10)
prf1 = reg1.params[0] + reg1.params[1]*x
ax.plot(x, prf1, color = 'red', label = 'Estimated PRF incl. Malta')
prf2 = reg2.params[0] + reg2.params[1]*x
ax.plot(x, prf2, color = 'blue', label = 'Estimated PRF excl. Malta')

ax.legend(loc='lower right')

plt.show()
```
