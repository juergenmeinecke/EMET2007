We will also again use the 'homemade' t-test function from Week 4. Don't forget you need to run the cell below before you can use the function:

```python
def t_test(x1, x2):
    numerator = x1.mean() - x2.mean()  # aka point estimate
    se1 = x1.std() / np.sqrt(len(x1))
    se2 = x2.std() / np.sqrt(len(x2))
    denominator = np.sqrt(se1**2 + se2**2)
    t_stat = numerator / denominator  # our t statistic
    ci_lb = numerator - 1.96 * denominator  # lower bound
    ci_ub = numerator + 1.96 * denominator  # upper bound
    ci = (ci_lb, ci_ub)
    print('Two-sample t-test')
    print(f'Mean in group 1: {x1.mean()}')
    print(f'Mean in group 2: {x2.mean()}')
    print(f'Point estimate for difference in means: {numerator}')
    print(f'Test statistic: {t_stat}')
    print(f'95% confidence interval: {ci}')
    return numerator, t_stat, ci
```

## Exercise 2

Construct a 95% confidence interval for the difference in `birthweight` between moms who smoke and moms who do not smoke.


```python
smokers = df[df.smoker == 1]
nonsmokers = df[df.smoker == 0]
t_test(smokers.birthweight, nonsmokers.birthweight)
```

### Exercise 3a
Run a simple regression of ``birthweight`` on ``smoker``.


```python
reg1 = smf.ols('birthweight ~ smoker', 
               data=df, 
               missing='drop').fit(cov_type='HC1', use_t=False)
reg1.summary()
```

## Exercise 4 
Do you think smoking is uncorrelated with other factors that cause low birthweight? How might this bias your estimate of 
$\beta_1$?


Smoking is a **choice** variable rather than an **assigned** variable; so it may be correlated with e.g. drinking, education, socio-economic background - all of which may be correlated with smoking *and* have an effect on birthweight.

In econometric analysis we need to be very careful generally of such associations, especially when working with 'choice' variables, as they lead to omitted variable bias (thus biasing our regression)! 

Therefore, in this case we have observed some omitted variable bias.  How do we deal with this?  In Ex 5 we run a multiple regression to control for some of these factors.


## Exercise 5 

### Exercise 5a
Run a multiple regression of ``birthweight`` on ``smoker``, ``alcohol``, and ``nprevist``.


```python
formula = 'birthweight ~ smoker + alcohol + nprevist'
reg2 = smf.ols(formula,
               data=df,
               missing='drop').fit(cov_type='HC1', use_t=False)
reg2.summary()
```

**Observations:**
- Coefficient estimate for smoker is now lower, confirming our suspicions that the original regression was subject to omitted variable bias
- The coefficient estimate on alcohol is not significant

### Exercise 5b

Construct a 95% confidence interval for $\beta_1$, that is, the effect of ``smoker`` on ``birthweight``. Is it substantively different from the regression that excludes ``alcohol`` and ``nprevist``?

```python
reg2.conf_int()
```

## Exercise 6

An alternative way to control for prenatal visits is to use the dummy variables ``tripre0`` through ``tripre3``. Notice that these four dummies are mutually exclusive while also being exhaustive of all possibilities with regards to prenatal visits: 

| Variable | Description |
| :-- | :-- |
| ``tripre0`` | Dummy equal to one if no pre-natal care visits, zero else |
| ``tripre1`` | Dummy equal to one if 1st pre-natal care visit in 1st trimester, zero else |
| ``tripre2`` | Dummy equal to one if 1st pre-natal care visit in 2nd trimester, zero else |
| ``tripre3`` | Dummy equal to one if 1st pre-natal care visit in 3rd trimester, zero else |


Run and compare the following multiple regressions:
* ``birthweight`` on ``smoker``, ``alcohol``, ``tripre0``, ``tripre2``, and ``tripre3`` 
* ``birthweight`` on ``smoker``, ``alcohol``, ``tripre0``, ``tripre1``, and ``tripre3`` 

Report the coefficient estimate for ``tripre0``. What does it capture?
Report the coefficient estimates for ``tripre2`` and ``tripre3``. What do they capture?
Why can we leave one trimester dummy out in either regression without sacrificing any information?


```python
formula_reg3 = 'birthweight ~ smoker + alcohol + tripre0 + tripre2 + tripre3'
reg3 = smf.ols(formula_reg3, 
                data=df,
                missing='drop').fit(cov_type='HC1', use_t=False)
reg3.summary()
```

```python
formula_reg4 = 'birthweight ~ smoker + alcohol + tripre0 + tripre1 + tripre3'
reg4 = smf.ols(formula_reg4, 
                data=df,
                missing='drop').fit(cov_type='HC1', use_t=False)
reg4.summary()
```

## tripre0
For the first regression, the coefficient estimate is -697.969.

This captures the average difference between not attending any prenatal care visits at all and attending the first prenatal care visit in the first trimester.

For the second regression, the coefficient estimate is -597.131.

This captures the average difference between not attending any prenatal care visits at all and attending the first prenatal care visit in the second trimester.

## tripre2

For the first regression, the coefficient estimate is -100.837.

This captures the average difference between attending the first prenatal care visit in the second trimester and attending the first prenatal care visit in the first trimester.

tripre2 was excluded from the second regression.


## tripre3

For the first regression, the coefficient estimate is -136.955.

This captures the average difference between attending the first prenatal care visit in the third trimester and attending the first prenatal care visit in the first trimester.

For the second regression, the coefficient estimate is -36.118.  This captures the average difference between attending the first prenatal care visit in the third trimester and attending the first prenatal care visit in 
the second trimester.

## Rest of answer

We can leave one trimester dummy out in either regression without sacrificing any information, as each mother can fall into a maximum of one category for the four trimester dummies.  That is, only one of tripre0 through tripre3 can be equal to one, and the rest will be equal to zero.

Each included trimester dummy captures the estimated effect on birthweight of being in that category, relative to being in the excluded category.  For example, excluding tripre0 from a multiple regression would mean that the other
trimester dummies would estimate the effect of having attended the first prenatal care visit in trimester X, compared to not attending any visits at all.
    
Note that we cannot include all of the trimester dummies in a multiple regression as this would be an instance of perfect multicollinearity.
