## Research Question

**Does lead in the water supply increase infant mortality rates?**

Lead is toxic and was commonly used in water pipes in the early 20th century. The amount of lead that leaches into water depends on the water's acidity (pH level) - more acidic water (lower pH) leaches more lead.

We'll investigate whether the effect of lead pipes on infant mortality depends on pH level using **interaction terms**.

## Part 2: Building Up to Interactions

### Exercise 4: Three Regression Specifications

We'll estimate three models to understand the value of interaction terms.

**Model 1:** Simple regression
$$\text{infRate} = \beta_0 + \beta_1 \cdot \text{lead} + u$$

**Model 2:** Multiple regression (additive)
$$\text{infRate} = \beta_0 + \beta_1 \cdot \text{lead} + \beta_2 \cdot \text{ph} + u$$

**Model 3:** With interaction
$$\text{infRate} = \beta_0 + \beta_1 \cdot \text{lead} + \beta_2 \cdot \text{ph} + \beta_3 \cdot (\text{lead} \times \text{ph}) + u$$

**Task 4b:** Interpret the coefficients in Model 3.

The model with interaction can be written as:

$$\text{infRate} = \beta_0 + \beta_1 \cdot \text{lead} + \beta_2 \cdot \text{ph} + \beta_3 \cdot (\text{lead} \times \text{ph}) + u$$

This implies **two separate regression lines:**

| When lead = | Regression line |
|:------------|:----------------|
| 0 (no lead) | $\text{infRate} = \beta_0 + \beta_2 \cdot \text{ph}$ |
| 1 (lead) | $\text{infRate} = (\beta_0 + \beta_1) + (\beta_2 + \beta_3) \cdot \text{ph}$ |

So:
- $\beta_1$ = difference in **intercepts** between lead and no-lead cities
- $\beta_3$ = difference in **slopes** between lead and no-lead cities

### Exercise 5: Joint Test for Lead Effect

To test whether lead has ANY effect (either through the level or the interaction), we need a joint F-test:

$$H_0: \beta_1 = \beta_3 = 0$$

## Summary

### Key Concepts

**Interaction Terms:**
- Allow the effect of one variable to depend on another
- Formula: `y ~ x1*x2` includes x1, x2, and x1:x2
- Create different slopes and intercepts for different groups

**Interpretation with Binary × Continuous Interaction:**
- $\beta_1$ (on the dummy): difference in intercepts
- $\beta_3$ (on the interaction): difference in slopes
- Effect of dummy depends on the continuous variable

**Testing Joint Significance:**
- Use F-test to test if dummy has any effect
- Test: $H_0: \beta_1 = \beta_3 = 0$

### Key Finding

Lead pipes increase infant mortality, but the effect is stronger in cities with more acidic water (lower pH). This makes physical sense: acidic water leaches more lead from the pipes.
