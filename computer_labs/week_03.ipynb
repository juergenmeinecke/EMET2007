## Part 3: Bivariate Analysis

So far we've done **univariate** analysis (one variable at a time). Now we'll explore **bivariate** analysis - studying relationships between two variables.

The approach depends on the variable types:

| Variable Types | Analysis Method |
|:---------------|:----------------|
| Two categorical | Contingency table (crosstab) |
| One categorical + one numerical | Group-by analysis |
| Two numerical | Correlation/covariance + scatter plot |

### Exercise 7: Covariance and Correlation (Two Numerical Variables)

For two numerical variables, we can summarise their linear association with:

- **Covariance**: Measures how two variables move together (units: product of original units)
- **Correlation**: Standardised covariance, always between -1 and +1

| Correlation Value | Interpretation |
|:------------------|:---------------|
| -1 | Perfect negative linear relationship |
| 0 | No linear relationship |
| +1 | Perfect positive linear relationship |

**Important:** These measures only capture *linear* relationships!

Use `.cov()` and `.corr()` on a subset of columns:

```python
df[['var1', 'var2']].cov()
df[['var1', 'var2']].corr()
```

### Exercise 8: Scatter Plots

Numbers alone can be misleading! Always visualise your data. A scatter plot shows the relationship between two numerical variables.

Basic matplotlib syntax:

```python
fig, ax = plt.subplots()                   # Create figure and axes
ax.plot(df.x_var, df.y_var, '.')           # Plot points (. = dots)
ax.set_xlabel('X axis label')              # Label x-axis
ax.set_ylabel('Y axis label')              # Label y-axis
ax.set_title('Plot title')                 # Add title
plt.show()                                 # Display the plot
```
