## Part 2: Investigating Omitted Variable Bias

### What is Omitted Variable Bias?

**Omitted Variable Bias (OVB)** occurs when:
1. A variable is omitted from the regression
2. The omitted variable is correlated with both the dependent variable AND an included regressor

If both conditions hold, the estimated coefficient on the included regressor will be **biased** - it will capture both its own effect AND part of the effect of the omitted variable.

### Is Gender a Confounding Variable?

Let's investigate whether **gender** might be causing OVB in our earnings-height regression.

1. Gender to be correlated with earnings

## Summary

### Key Concepts

**Hypothesis Testing:**
- Three equivalent methods: t-statistic, confidence interval, p-value
- All should give the same conclusion

**Omitted Variable Bias:**
- Occurs when an omitted variable is correlated with both Y and an included X
- Biases the coefficient on the included variable
- Can lead to incorrect causal interpretations

**Subsample Analysis:**
- Running regressions on subsamples can help address certain types of OVB
- Useful when you suspect a categorical variable is confounding results

**Units of Measurement:**
- Changing units affects the magnitude of coefficients
- Does not affect statistical significance or model fit (R²)
- Slope scales inversely with the unit change
